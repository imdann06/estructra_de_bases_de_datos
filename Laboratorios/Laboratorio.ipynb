#Problema 2

Dado el hash root del arbol de merkel, determinar el orden de las transacciones que lo generan.

Reglas las transaciiones son conocidas y el numero de transacciones

```python
import hashlib
import itertools        # Permite generar permutaciones

def merkle_root(txs): # Convierte cada transacción en su hash SHA-256
    nivel = [hashlib.sha256(tx.encode()).hexdigest() for tx in txs]  # Esto crea el nivel hoja del árbol
    while len(nivel) > 1: # Repite hasta que solo quede un hash
        nuevo = [] # Aquí se guardará el siguiente nivel del árbol
        i = 0

        while i < len(nivel):  # Recorremos el nivel actual.

            if i + 1 < len(nivel): # Si existe un par, se combinan dos hashes
                combinado = (nivel[i] + nivel[i + 1]).encode()
                nuevo.append(hashlib.sha256(combinado).hexdigest()) # Calcula el hash del par y lo agrega al siguiente nivel

                i += 2 # Avanza dos posiciones

            else:

                nuevo.append(nivel[i])  # Si queda un hash sin pareja, sube igual al siguiente nivel

                i += 1

        nivel = nuevo # El siguiente nivel pasa a ser el nivel actual

    return nivel[0] # Cuando queda un solo hash, ese es el Merkle root


def encontrar_orden(transacciones, root_objetivo):

    for p in itertools.permutations(transacciones): # Genera todas las formas posibles de ordenar las transacciones

        root_calculado = merkle_root(p)  # Calcula el Merkle root de esa permutación

        if root_calculado == root_objetivo: # Compara el root calculado con el root objetivo
            return p # Si coincide, devuelve ese orden
    return None # Si termina todas las permutaciones y no coincide ninguna, devuelve None

n = int(input("Número de transacciones: ")) # Se piden el numero de transacciones
txs = []
for i in range(n):

    tx = input(f"Transacción {i+1}: ") # Pide una transacción y la guarda en la variable tx
    txs.append(tx) # Agrega la transacción a la lista
root_objetivo = input("Merkle root objetivo: ") # Pide el Merkle root que se quiere igualar
resultado = encontrar_orden(txs, root_objetivo) # Llama a la función que busca el orden correcto


if resultado:
    print("Orden correcto encontrado:")
    print(resultado)
else:
    print("No se encontró un orden que genere ese root")
```
