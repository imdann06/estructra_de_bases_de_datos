A partir de los experimentos realizados, se observa que el QuadTreecpresenta un mejor desempeño en el tiempo de consulta en comparación con la búsqueda por fuerza bruta, especialmente a medida que aumenta el tamaño del conjunto de datos.

Sin embargo, el QuadTree introduce un costo adicional asociado a su construcción. En los casos donde se realiza una única consulta o el tamaño de datos es pequeño, la fuerza bruta puede ser mas rapida,
