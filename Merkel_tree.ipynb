```python
import hashlib

texto1= "hola"
hash1= hashlib.sha256(texto1.encode())
print("hola:", hash1.hexdigest())

texto2= "Hola"
hash2= hashlib.sha256(texto2.encode())
print("Hola:", hash2.hexdigest())

texto3= "chao"
hash3= hashlib.sha256(texto3.encode())
print("chao:", hash3.hexdigest())

texto4= "Chao"
hash4= hashlib.sha256(texto4.encode())
print("Chao:", hash4.hexdigest())

holas_combined_string = hash1.hexdigest() + hash2.hexdigest()
print("Holas (combined hexdigest strings):", holas_combined_string)

chao_combined_string = hash3.hexdigest() + hash4.hexdigest()
print("Chao (combined hexdigest strings):", chao_combined_string)

hash_hola= hashlib.sha256(holas_combined_string.encode())
print("Hash de Holas:", hash_hola.hexdigest())

hash_chao= hashlib.sha256(chao_combined_string.encode())
print("Hash de Chao:", hash_chao.hexdigest())

hash_final = hashlib.sha256(hash_hola.hexdigest().encode() + hash_chao.hexdigest().encode())
print("Hash Final:", hash_final.hexdigest())
```

```python
#Arbol con 4 entradas

texto1,texto2,texto3,texto4= "hola", "Hola", "chao", "Chao"

def hash_string(string):
    return hashlib.sha256(string.encode()).hexdigest()

def hash_tree(left, right):
    return hash_string(left + right)

def hash_tree_4(a, b, c, d):
    hash1 = hash_string(a)
    print(f"Hash 1: {a}: {hash1}")
    hash2 = hash_string(b)
    print(f"Hash 2: {b}: {hash2}")
    hash3 = hash_string(c)
    print(f"Hash 3: {c}: {hash3}")
    hash4 = hash_string(d)
    print(f"Hash 4: {d}: {hash4}")

    left_hash = hash_tree(hash1, hash2)
    print("Left Hash:", left_hash)
    right_hash = hash_tree(hash3, hash4)
    print("Right Hash:", right_hash)

    return hash_tree(left_hash, right_hash)

hash_result = hash_tree_4(texto1, texto2, texto3, texto4)
print("Hash Final:", hash_result)
```

```python
#Arbol con 5 entradas
n1,n2,n3,n4,n5= "hola", "Hola", "chao", "Chao", "Hello"

def hash_string(string):
    return hashlib.sha256(string.encode()).hexdigest()

def hash_tree(left, right):
    return hash_string(left + right)

def hash_tree_5(a, b, c, d, e):
    hash1 = hash_string(a)
    print(f"Hash 1: {a}: {hash1}")
    hash2 = hash_string(b)
    print(f"Hash 2: {b}: {hash2}")
    hash3 = hash_string(c)
    print(f"Hash 3: {c}: {hash3}")
    hash4 = hash_string(d)
    print(f"Hash 4: {d}: {hash4}")
    hash5 = hash_string(e)
    print(f"Hash 5: {e}: {hash5}")
    left_hash = hash_tree(hash1, hash2)
    print("Left Hash:", left_hash)
    right_hash = hash_tree(hash3, hash4)
    print("Right Hash:", right_hash)
    root= hash_tree(left_hash, right_hash) + hash5
    root = hash_string(root)
    print("Root Hash:", root)
    return root

hash_result = hash_tree_5(n1, n2, n3, n4, n5)
```
